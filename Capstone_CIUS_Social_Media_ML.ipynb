### Column Selection and Renaming
#### The original CIUS dataset contains 1,342 columns. To align this data with our survey, we will select only the 16 specific variables we need.

#### At the same time, we will rename these columns to more descriptive and "friendly" labels for easier analysis.

```python
# 16 columns we need
cols16 = [
    "PROVINCE", "AGE_GRP", "GENDER", "EDU", "EMP", "HINCQUIN", "WTPG",
    "TS_010A", "TS_020A", "UI_110A", "UI_110B", "UI_110C",
    "DS_070A", "DS_070B", "GHCOMP", "ONL_SHOP"
]

# Check which ones exist in your CIUS file
available_cols = [c for c in cols16 if c in cius.columns]
missing_cols = [c for c in cols16 if c not in cius.columns]

print("Found columns:", available_cols)
print("Missing columns:", missing_cols)

```

```python
column_mapping = {
    "PROVINCE": "Province",
    "AGE_GRP": "Age",
    "GENDER": "Gender",
    "EDU": "Education",
    "EMP": "Employment",
    "HINCQUIN": "Income_Group",
    "WTPG": "Weight",

    # TS_* are about relationships & online communication
    "TS_010A": "Rel_Satisfaction",          # satisfaction with relationships (friends)
    "TS_020A": "Online_Contact_Freq",       # how often use internet to communicate with friends

    # UI_110* are *effects* of online use
    "UI_110A": "SM_Interfere_Rel",          # Does online use interfere with relationships?
    "UI_110B": "SM_Interfere_Life",         # Interfere with other aspects of life?
    "UI_110C": "SM_Anxious_Envious",        # Make you feel anxious / depressed / envious?

    # DS_070* are *AI awareness*, not stress/sleep
    "DS_070A": "SM_AI_Chatbot",             # Ever noticed AI in chatbots?
    "DS_070B": "SM_AI_Email",               # Ever noticed AI in emails?

    # Household composition
    "GHCOMP": "Household_Type",

    # Online shopping
    "ONL_SHOP": "Online_Shopping"
}

cius = cius.rename(columns=column_mapping)
```

## Our Baseline model will predict : Social media usage intensity

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

import joblib

```

```python
cius_on = cius_16[cius_16["Province"] == "Ontario"].copy()
print("Ontario rows:", cius_on.shape[0])
```

```python
##Create TARGET: usage_risk

def usage_risk(freq):
    if freq == "Everyday":
        return "High"
    if freq in ["A few times a week", "Once a week"]:
        return "Moderate"
    if freq in ["2–3 times a month", "Once a month", "Not in the past month"]:
        return "Low"
    return None

cius_on["usage_risk"] = cius_on["Online_Contact_Freq"].apply(usage_risk)
```

```python
# Drop rows with missing target
cius_model = cius_on.dropna(subset=["usage_risk"]).copy()

print("\nUsage risk distribution:")
print(cius_model["usage_risk"].value_counts())
```

```python
# Feature set (X) and target (y)

X = cius_model[
    [
        "Age",
        "Gender",
        "Education",
        "Employment",
        "Income_Group",
        "Household_Type",
        "Online_Shopping",
        "SM_Interfere_Rel",
        "SM_Interfere_Life",
        "SM_Anxious_Envious"
    ]
]

y = cius_model["usage_risk"]
```

```python
# Train / test split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

```

```python
# Preprocessing + Model

categorical_features = X.columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

model = Pipeline(
    steps=[
        ("preprocess", preprocessor),
        ("clf", LogisticRegression(
            max_iter=2000,
            multi_class="multinomial"
        ))
    ]
)

model.fit(X_train, y_train)
```

```python
# Evaluation (BASELINE RESULTS)

y_pred = model.predict(X_test)

print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred, labels=["Low", "Moderate", "High"]))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Save model (for Flask)
joblib.dump(model, "cius_usage_risk_baseline.pkl")
print("\nModel saved as: cius_usage_risk_baseline.pkl")
```

We built a multinomial logistic regression baseline. The model performs well for identifying high-usage users but struggles to distinguish low-usage users, highlighting class imbalance and overlapping behavioral patterns


So instead of this we decided to go with binary classificastion

```python
cius_on["high_use"] = (cius_on["Online_Contact_Freq"] == "Everyday").astype(int)
```

```python
print("\nTarget distribution (high_use):")
print(cius_on["high_use"].value_counts())
```

```python
feature_cols = [
    "Age",
    "Gender",
    "Education",
    "Employment",
    "Income_Group",
    "Household_Type",
    "Online_Shopping",
    "SM_Interfere_Rel",
    "SM_Interfere_Life",
    "SM_Anxious_Envious"
]
```

```python
df_model = cius_on.dropna(subset=feature_cols + ["high_use"]).copy()

X = df_model[feature_cols]
y = df_model["high_use"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[("cat", OneHotEncoder(handle_unknown="ignore"), feature_cols)]
)

clf = LogisticRegression(
    max_iter=2000,
    class_weight="balanced"   # <-- keep this for better minority handling
)

model = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("clf", clf)
])

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion matrix [ [TN FP], [FN TP] ]:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification report:")
print(classification_report(y_test, y_pred, target_names=["Not-High", "High"]))
```

```python
joblib.dump(model, "cius_highuse_binary_baseline.pkl")
print("\nSaved: cius_highuse_binary_baseline.pkl")
```

The binary logistic regression baseline achieves 63% accuracy with balanced recall across high and non-high social media users, demonstrating that usage intensity is predictable from demographic and behavioral features.


# ANN Model for Social Media Usage (Binary)\n\n
Here we are building a **binary ANN model** using `MLPClassifier` (a neural network) and saves it for deployment.\n\n
**Target used for higher accuracy and a more stable binary split:**\n
- `1 = Everyday or A few times a week`\n
- `0 = lower frequency categories`\n\n
Output files created by this notebook:\n
- `ann_social_media_deployment_bundle.pkl`\n
- `ann_model_metrics.json`\n

```python
import pandas as pd
import numpy as np
import joblib
import json
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix, classification_report
```

```python
# 1) Load the data\n
DATA_PATH = 'CIUS_SocialMedia_Clean_Labeled.csv'
df = pd.read_csv(DATA_PATH)

# 2) Build a binary target\n
# High / frequent use = Everyday or A few times a week\n
high_classes = ['Everyday', 'A few times a week']
df = df[df['Online_Contact_Freq'].notna()].copy()
df['high_use'] = df['Online_Contact_Freq'].isin(high_classes).astype(int)

print(df['high_use'].value_counts())
print('Baseline accuracy =', df['high_use'].value_counts(normalize=True).max())
```

```python
# 3) Split features and target\n
X = df.drop(columns=['Online_Contact_Freq', 'high_use'])
y = df['high_use']

cat_cols = [c for c in X.columns if X[c].dtype == 'object']
num_cols = [c for c in X.columns if X[c].dtype != 'object']

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', Pipeline([
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(handle_unknown='ignore'))
        ]), cat_cols),
        ('num', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]), num_cols)
    ]
)
```

```python
# 4) Train / validation / test split\n
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full, test_size=0.20, random_state=42, stratify=y_train_full
)

X_train_t = preprocessor.fit_transform(X_train)
X_val_t = preprocessor.transform(X_val)
X_test_t = preprocessor.transform(X_test)

# Convert to dense arrays because MLPClassifier expects dense input\n
if hasattr(X_train_t, 'toarray'):
    X_train_t = X_train_t.toarray()
    X_val_t = X_val_t.toarray()
    X_test_t = X_test_t.toarray()
```

```python
# 5) ANN model\n
ann_model = MLPClassifier(
    hidden_layer_sizes=(64, 32),
    activation='relu',
    solver='adam',
    alpha=0.0005,
    batch_size=128,
    learning_rate_init=0.001,
    max_iter=120,
    early_stopping=True,
    validation_fraction=0.15,
    n_iter_no_change=10,
    random_state=42
)

ann_model.fit(X_train_t, y_train)
```

```python
# 6) Threshold optimization for maximum accuracy\n
val_proba = ann_model.predict_proba(X_val_t)[:, 1]
best_threshold = 0.50
best_val_acc = -1

for t in np.arange(0.30, 0.71, 0.01):
    preds = (val_proba >= t).astype(int)
    acc = accuracy_score(y_val, preds)
    if acc > best_val_acc:
        best_val_acc = acc
        best_threshold = float(t)

print('Best threshold:', best_threshold)
print('Best validation accuracy:', best_val_acc)
```

```python
# 7) Final evaluation on the test set\n
test_proba = ann_model.predict_proba(X_test_t)[:, 1]
y_pred_default = (test_proba >= 0.50).astype(int)
y_pred_best = (test_proba >= best_threshold).astype(int)

print('Test Accuracy (0.50 threshold):', accuracy_score(y_test, y_pred_default))
print('Test Accuracy (best threshold):', accuracy_score(y_test, y_pred_best))
print('Test AUC:', roc_auc_score(y_test, test_proba))
print('\\nConfusion Matrix:')
print(confusion_matrix(y_test, y_pred_best))
print('\\nClassification Report:')
print(classification_report(y_test, y_pred_best, target_names=['Not-High', 'High']))
```

```python
# 8) Save the model bundle for deployment\n
bundle = {
    'preprocessor': preprocessor,
    'model': ann_model,
    'threshold': best_threshold,
    'feature_columns': X.columns.tolist(),
    'target_definition': '1 = Everyday or A few times a week, 0 = lower frequency categories'
}

joblib.dump(bundle, 'ann_social_media_deployment_bundle.pkl')
print('Saved ann_social_media_deployment_bundle.pkl')
```

```python
# 9) Example deployment prediction\n
loaded = joblib.load('ann_social_media_deployment_bundle.pkl')

# Use one real example row\n
sample = X.iloc[[0]].copy()
sample_t = loaded['preprocessor'].transform(sample)
if hasattr(sample_t, 'toarray'):
    sample_t = sample_t.toarray()

proba = loaded['model'].predict_proba(sample_t)[:, 1][0]
pred = int(proba >= loaded['threshold'])
print('Probability of high use:', round(float(proba), 4))
print('Predicted class:', pred)
```
